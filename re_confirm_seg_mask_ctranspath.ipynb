```python
def load_ctranspath_model(model_path: str):
    print("⚡ 모델 로드 및 가중치 파일 확인 중...")
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"🔥 사용 장치: {device}")

    if not os.path.exists(model_path):
        raise FileNotFoundError(f"모델 파일을 찾을 수 없습니다: {model_path}")

    model = timm.create_model("swin_tiny_patch4_window7_224", pretrained=False)
    state_dict = torch.load(model_path, map_location=device)

    if "model" in state_dict:
        state_dict = state_dict["model"]
    elif "state_dict" in state_dict:
        state_dict = state_dict["state_dict"]

    model.load_state_dict(state_dict, strict=False)
    model.head = nn.Identity()
    model.to(device)
    model.eval()
    print("🎉 CTransPath 모델 로드 성공")
    return model, device


model, device = load_ctranspath_model(MODEL_PATH)
```

Transformers 설정

 이미지를 모델이 원하는 $224 \times 224$ 크기로 강제 변환합니다.

```python
def build_val_transform():
    return transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
    ])


val_transform = build_val_transform()
print("✅ 전처리 도구 설정 완료 (224x224 Resize 포함)")
```

WSI 슬라이드 정보 읽기

슬라이드 전체 크기를 확인하고 분석할 레벨(해상도)을 정합니다.

```python
def open_wsi_at_level(wsi_path: str, requested_level: int):
    print("🔍 슬라이드 정보 읽는 중...")
    reader = WSIReader.open(wsi_path)
    info = reader.info

    if requested_level < len(info.level_dimensions):
        level = requested_level
        level_w, level_h = info.level_dimensions[level]
        scale_factor = info.level_downsamples[level]
    else:
        print("⚠ Level 정보가 부족합니다. Level 0을 기준으로 강제 축소(4배)하여 진행합니다.")
        level = 0
        scale_factor = 4.0
        level_w = int(info.level_dimensions[0][0] / scale_factor)
        level_h = int(info.level_dimensions[0][1] / scale_factor)

    print(f"- 선택된 Level {level} 크기: {level_w} x {level_h}")
    print(f"- 축소 비율: {scale_factor:.1f}배 (Level 0 대비)")
    return reader, info, level, scale_factor, level_w, level_h


reader, info, target_level, scale_factor, level_w, level_h = open_wsi_at_level(WSI_PATH, TARGET_LEVEL)
```

패치 추출 및 특징 추출

이미지를 조각내어 모델에 넣고 특징을 뽑아냅니다. (시간이 좀 걸릴 수 있습니다)

```python
def extract_patch_features(reader, model, device, transform, level, level_w, level_h, scale_factor, patch_size, batch_size):
    features_list = []
    coords_list = []
    xs = range(0, level_w, patch_size)
    ys = range(0, level_h, patch_size)
    total_patches = len(xs) * len(ys)

    print(f"🚀 총 {total_patches}개 패치 분석 시작...")
    batch_imgs, batch_coords = [], []

    def flush_batch():
        if not batch_imgs:
            return
        batch_tensor = torch.stack(batch_imgs).to(device)
        feats = model(batch_tensor)
        features_list.append(feats.cpu().numpy())
        coords_list.extend(batch_coords)
        batch_imgs.clear()
        batch_coords.clear()

    with torch.no_grad():
        for y in tqdm(ys):
            for x in xs:
                location = (int(x * scale_factor), int(y * scale_factor))
                try:
                    patch = reader.read_region(location=location, level=level, size=(patch_size, patch_size))
                except Exception:
                    continue

                patch = np.array(patch)
                if patch.shape[-1] == 4:
                    patch = patch[:, :, :3]

                batch_imgs.append(transform(Image.fromarray(patch)))
                batch_coords.append((x, y))

                if len(batch_imgs) >= batch_size:
                    flush_batch()

        flush_batch()

    print("\n✅ 모든 패치 분석 완료!")
    return features_list, coords_list


features_list, coords_list = extract_patch_features(
    reader=reader,
    model=model,
    device=device,
    transform=val_transform,
    level=target_level,
    level_w=level_w,
    level_h=level_h,
    scale_factor=scale_factor,
    patch_size=PATCH_SIZE,
    batch_size=BATCH_SIZE,
)
```

클러스터링 (K-Means)

추출된 특징들을 분석하여 5가지(설정 가능) 조직 유형으로 자동 분류합니다.

```python
def cluster_features(features_list, coords_list, n_clusters, reference_centroids_file, level_w, level_h, patch_size):
    if not features_list:
        raise ValueError("추출된 특징이 없습니다. WSI 경로와 패치 추출 설정을 확인하세요.")

    all_features = np.concatenate(features_list, axis=0)
    all_coords = np.array(coords_list)
    print(f"📊 클러스터링 데이터 크기: {all_features.shape}")

    if os.path.exists(reference_centroids_file):
        reference_centroids = np.load(reference_centroids_file)
        if reference_centroids.shape[0] != n_clusters:
            raise ValueError(
                f"저장된 centroid 개수({reference_centroids.shape[0]})와 N_CLUSTERS({n_clusters})가 다릅니다."
            )
        labels = pairwise_distances_argmin(all_features, reference_centroids)
        print("✅ 저장된 centroid 기준으로 고정 분류 완료")
    else:
        print("⚠️ 참조 centroid가 없어 현재 데이터로 새 기준을 생성합니다.")
        kmeans = KMeans(n_clusters=n_clusters, random_state=42, n_init=10)
        labels = kmeans.fit_predict(all_features)
        np.save(reference_centroids_file, kmeans.cluster_centers_)
        print(f"💾 생성된 centroid 저장 완료: {reference_centroids_file}")
        print("📌 저장 후에는 Cluster ID(0~4)와 조직 유형 매핑을 한 번 확인하세요.")

    grid_w = int(np.ceil(level_w / patch_size))
    grid_h = int(np.ceil(level_h / patch_size))
    segmentation_map = np.full((grid_h, grid_w), -1, dtype=np.int16)

    for (x, y), label in zip(all_coords, labels):
        gx, gy = int(x / patch_size), int(y / patch_size)
        if 0 <= gy < grid_h and 0 <= gx < grid_w:
            segmentation_map[gy, gx] = label

    print(f"🎉 세그멘테이션 맵 생성 완료! 크기: {segmentation_map.shape}")
    return all_features, all_coords, labels, segmentation_map


all_features, all_coords, labels, segmentation_map = cluster_features(
    features_list=features_list,
    coords_list=coords_list,
    n_clusters=N_CLUSTERS,
    reference_centroids_file=REFERENCE_CENTROIDS_FILE,
    level_w=level_w,
    level_h=level_h,
    patch_size=PATCH_SIZE,
)
```

결과 시각화

원본 이미지와 분석 결과(색상 맵)를 겹쳐서 보여줍니다.

```python
def render_segmentation_overlay(reader, target_level, segmentation_map, n_clusters, colors):
    custom_cmap = mpl.colors.ListedColormap(colors[:n_clusters])
    w, h = reader.info.level_dimensions[target_level]

    print(f"🎨 시각화 생성 중 (Level {target_level})...")
    try:
        high_res_img = np.array(reader.read_region((0, 0), target_level, (w, h)))[:, :, :3]
    except Exception as exc:
        print(f"⚠ 전체 이미지 읽기 실패: {exc}")
        high_res_img = np.zeros((h, w, 3), dtype=np.uint8)

    resized_map = cv2.resize(
        segmentation_map.astype(np.float32),
        dsize=(w, h),
        interpolation=cv2.INTER_NEAREST,
    )

    fig, axes = plt.subplots(1, 2, figsize=(20, 10))
    axes[0].imshow(high_res_img)
    axes[0].set_title(f"Original Image (Level {target_level})", fontsize=15)
    axes[0].axis("off")

    axes[1].imshow(high_res_img)
    overlay = axes[1].imshow(resized_map, cmap=custom_cmap, alpha=0.6, vmin=0, vmax=n_clusters - 1)
    axes[1].set_title("AI Analysis Result (Segmentation Overlay)", fontsize=15)
    axes[1].axis("off")

    cbar = fig.colorbar(overlay, ax=axes[1], fraction=0.046, pad=0.04)
    cbar.set_label("Cluster ID", fontsize=12)
    cbar.set_ticks(np.arange(0, n_clusters, 1))

    plt.tight_layout()
    plt.show()
    print("🎉 시각화 처리가 완료되었습니다.")
    return high_res_img, resized_map, custom_cmap


high_res_img, resized_map, custom_cmap = render_segmentation_overlay(
    reader=reader,
    target_level=target_level,
    segmentation_map=segmentation_map,
    n_clusters=N_CLUSTERS,
    colors=CUSTOM_COLORS,
)
```

```python
def print_lesion_summary(resized_map, cluster_definitions, major_lesion_ids):
    unique_ids, counts = np.unique(resized_map, return_counts=True)
    pixel_counts = dict(zip(unique_ids.astype(int), counts))
    total_tissue_pixels = sum(count for id_val, count in pixel_counts.items() if id_val != 0)

    if total_tissue_pixels == 0:
        print("❌ 분석할 조직 영역이 발견되지 않았습니다.")
        return None

    major_lesion_pixels = sum(pixel_counts.get(id_val, 0) for id_val in major_lesion_ids)
    lesion_percentage = (major_lesion_pixels / total_tissue_pixels) * 100
    summary_sentence = f"판독 결과: 전체 조직 중 주요 병변 영역은 약 {lesion_percentage:.1f}%를 차지하고 있는 것으로 확인되었습니다."

    print("-" * 60)
    print("📊 [AI 이미지 분석 결과 보고서]")
    print("1. 분석 대상: 조직 내 주요 분포 확인")
    print(f"2. 주요 병변 점유율: {lesion_percentage:.1f}% (배경 제외 기준)")
    print("-" * 60)
    print("📢 환자를 위한 판독 결과")
    print(summary_sentence)
    print("-" * 60)
    print("\n[참고: 상세 분포]")

    for cluster_id in range(1, len(cluster_definitions)):
        pct = (pixel_counts.get(cluster_id, 0) / total_tissue_pixels) * 100
        print(f"- {cluster_definitions[cluster_id]}: {pct:.1f}%")

    return lesion_percentage


lesion_percentage = print_lesion_summary(resized_map, CLUSTER_DEFINITIONS, MAJOR_LESION_IDS)
```

```python
def build_cluster_mask_overlay(high_res_img, resized_map, target_cluster_ids, overlay_color, overlay_alpha):
    binary_mask = np.isin(resized_map, target_cluster_ids)
    color_mask = np.zeros((*binary_mask.shape, 4), dtype=np.uint8)
    color_mask[binary_mask, :3] = overlay_color
    color_mask[binary_mask, 3] = int(overlay_alpha * 255)

    fig, axes = plt.subplots(1, 3, figsize=(24, 8))
    axes[0].imshow(high_res_img)
    axes[0].set_title(f"1. Original Image (Level {target_level})", fontsize=14, fontweight="bold")
    axes[0].axis("off")

    axes[1].imshow(high_res_img)
    axes[1].imshow(resized_map, cmap=custom_cmap, alpha=0.6, vmin=0, vmax=N_CLUSTERS - 1)
    axes[1].set_title("2. All Clusters Segmentation Overlay", fontsize=14, fontweight="bold")
    axes[1].axis("off")

    axes[2].imshow(high_res_img)
    axes[2].imshow(color_mask)
    axes[2].set_title(f"3. Combined Mask (IDs: {target_cluster_ids})", fontsize=14, fontweight="bold")
    axes[2].axis("off")

    plt.tight_layout()
    plt.show()
    print(f"🎉 특정 클러스터 ID {target_cluster_ids} 마스크가 통합 표시되었습니다.")
    return (binary_mask.astype(np.uint8) * 255), color_mask


binary_mask, color_mask = build_cluster_mask_overlay(
    high_res_img=high_res_img,
    resized_map=resized_map,
    target_cluster_ids=TARGET_CLUSTER_IDS,
    overlay_color=OVERLAY_COLOR,
    overlay_alpha=OVERLAY_ALPHA,
)
```

```python
def save_visualizations(wsi_path, output_dir, target_level, high_res_img, resized_map, custom_cmap, n_clusters, color_mask, target_cluster_ids):
    file_name_base = os.path.splitext(os.path.basename(wsi_path))[0]
    save_dir = os.path.join(output_dir, "visualizations")
    os.makedirs(save_dir, exist_ok=True)
    print(f"📂 저장 폴더: {save_dir}")

    original_path = os.path.join(save_dir, f"{file_name_base}_1_Original_L{target_level}.jpg")
    Image.fromarray(high_res_img).save(original_path)
    print(f"✅ 원본 이미지 저장 완료: {original_path}")

    norm = mpl.colors.Normalize(vmin=0, vmax=n_clusters - 1)
    mapper = mpl.cm.ScalarMappable(norm=norm, cmap=custom_cmap)
    seg_rgb_uint8 = (mapper.to_rgba(resized_map)[:, :, :3] * 255).astype(np.uint8)
    seg_overlay_img = cv2.addWeighted(seg_rgb_uint8, 0.6, high_res_img, 0.4, 0)

    seg_path = os.path.join(save_dir, f"{file_name_base}_2_SegmentationOverlay_L{target_level}.jpg")
    Image.fromarray(seg_overlay_img).save(seg_path)
    print(f"✅ 세그멘테이션 오버레이 저장 완료: {seg_path}")

    original_pil = Image.fromarray(high_res_img)
    mask_pil = Image.fromarray(color_mask, "RGBA")
    mask_overlay_pil = original_pil.copy()
    mask_overlay_pil.paste(mask_pil, (0, 0), mask_pil)

    ids_str = "_".join(map(str, target_cluster_ids))
    mask_path = os.path.join(save_dir, f"{file_name_base}_3_CombinedMask_IDs_{ids_str}_L{target_level}.jpg")
    mask_overlay_pil.save(mask_path)
    print(f"✅ 통합 마스크 오버레이 저장 완료: {mask_path}")
    print("\n🎉 모든 시각화 결과 저장이 완료되었습니다.")
    return original_path, seg_path, mask_path


saved_paths = save_visualizations(
    wsi_path=WSI_PATH,
    output_dir=OUTPUT_DIR,
    target_level=target_level,
    high_res_img=high_res_img,
    resized_map=resized_map,
    custom_cmap=custom_cmap,
    n_clusters=N_CLUSTERS,
    color_mask=color_mask,
    target_cluster_ids=TARGET_CLUSTER_IDS,
)
```
